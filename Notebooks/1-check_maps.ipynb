## Mapas que dependen de la fijación

Corrido con **python3 extract_human_maps.py --d HSEM --cfg elm_final**

Volví a correr el modelo para obtener los mapas de las fijaciones de los humanos (forcé el modelo a que mire a donde miró el humano)

```python
# ── Leer mapas de un sujeto e imagen específicos ──────────────────────────────
from pathlib import Path

MAPS_DIR = Path('../Human_Maps/HSUBANOTT')

def cargar_mapas(subject_id, image_name):
    """
    Carga los mapas extraídos para una fijación de un sujeto.
    
    Retorna un dict con:
      visual_evidence : (n_fix, 24, 32)  — evidencia visual W
      posterior       : (n_fix, 24, 32)  — probabilidad posterior
      entropy         : (n_fix,)         — entropía del posterior
      fixations_y     : (n_fix,)         — coord Y en grilla
      fixations_x     : (n_fix,)         — coord X en grilla
      memory_set, target_stim, target_found
    """
    npz_name = image_name.replace('.jpg', '').replace('.png', '') + '.npz'
    path = MAPS_DIR / subject_id / npz_name
    data = np.load(path, allow_pickle=True)
    return dict(data)

# Ejemplo de uso
maps = cargar_mapas('et_117969', 'cmp_building_018_cat_005.jpg')

print('n_fijaciones:', maps['posterior'].shape[0])
print('entropy por fijación:', maps['entropy_map'])
print('target_found:', maps['target_found'])
```

```python
# ── Acceder a un mapa en una fijación específica ───────────────────────────────
maps = cargar_mapas('et_117969', 'cmp_building_018_cat_005.jpg')

fixacion = 3  # índice (0-based)

posterior_fix3   = maps['posterior'][fixacion]        # (24, 32)
evidencia_fix3   = maps['visual_evidence'][fixacion]  # (24, 32)
entropia_fix3    = maps['entropy_map'][fixacion]          # escalar
coord_y, coord_x = maps['fixations_y'][fixacion], maps['fixations_x'][fixacion]

print(f'Fijación {fixacion}: celda ({coord_y}, {coord_x}), entropía={entropia_fix3}')
print(f'Posterior suma: {posterior_fix3.sum():.4f}')  # debe ser ~1.0
```

```python
# ── Iterar sobre todos los sujetos ────────────────────────────────────────────
def listar_sujetos():
    return sorted([p.name for p in MAPS_DIR.iterdir() if p.is_dir()])

def listar_imagenes(subject_id):
    return sorted([p.stem + '.jpg' for p in (MAPS_DIR / subject_id).glob('*.npz')])

# Ver sujetos disponibles
sujetos = listar_sujetos()
print(f'{len(sujetos)} sujetos:', sujetos[:3], '...')

# Ver imágenes de un sujeto
imagenes = listar_imagenes('et_117969')
print(f'{len(imagenes)} imágenes para et_117969')
```

```python
### CHECK 2: Grafico la posicion de las fijaciones en la imagen

def graficar_fijaciones(maps, imagen):
    import matplotlib.pyplot as plt
    from PIL import Image
    
    img_path = Path('../Datasets/HSEM/images/') / imagen
    img = Image.open(img_path)
    img_w, img_h = img.size

    grid_h, grid_w = maps['posterior'].shape[-2:]

    fix_x = [(x + 0.5) * (img_w / grid_w) for x in maps['fixations_x']]
    fix_y = [(y + 0.5) * (img_h / grid_h) for y in maps['fixations_y']]
    
    plt.figure(figsize=(8, 6))
    plt.imshow(img)
    plt.scatter(fix_x, fix_y, c='red', s=100, edgecolors='white', label='Fijaciones')
    
    for i, (x, y) in enumerate(zip(fix_x, fix_y)):
        plt.text(x, y, str(i), color='white', fontsize=8, ha='center', va='center', weight='bold')
    
    plt.title(f'Fijaciones para {imagen}')
    plt.legend()
    plt.axis('off')
    plt.show()

graficar_fijaciones(maps, imagen)
```

```python
### CHECK 3: Miro mapa de entropia para la primera fijacion

def graficar_media_entropia_por_fijacion(maps):
    entropias = maps['entropy_map'].mean(axis=(1, 2))
    plt.figure(figsize=(8, 4))
    plt.plot(entropias, marker='o')
    plt.title('Media Entropía del posterior por fijación')
    plt.xlabel('Número de fijación')
    plt.ylabel('Media Entropía')
    plt.grid()
    plt.show()

graficar_media_entropia_por_fijacion(maps)
```

```python
def graficar_mapas_por_fijacion(maps, imagen, fixacion):
    import matplotlib.pyplot as plt
    from PIL import Image
    import numpy as np

    img_path = Path('../Datasets/HSEM/images/') / imagen
    img = Image.open(img_path)
    img_w, img_h = img.size

    grid_h, grid_w = maps['posterior'].shape[-2:]

    fix_x = (maps['fixations_x'][fixacion] + 0.5) * (img_w / grid_w)
    fix_y = (maps['fixations_y'][fixacion] + 0.5) * (img_h / grid_h)

    posterior = maps['posterior'][fixacion]
    evidencia = maps['visual_evidence'][fixacion]
    entropia = maps['entropy_map'][fixacion]

    fig, axs = plt.subplots(1, 4, figsize=(22, 5))

    # Imagen con fijación
    axs[0].imshow(img)
    axs[0].scatter(fix_x, fix_y, c='red', s=150, edgecolors='white', zorder=5)
    axs[0].text(fix_x, fix_y, str(fixacion), color='white', fontsize=8,
                ha='center', va='center', weight='bold', zorder=6)
    axs[0].set_title(f'Imagen - Fijación {fixacion}')

    im1 = axs[1].imshow(posterior, cmap='viridis')
    axs[1].set_title(f'Posterior - Fijación {fixacion}')
    plt.colorbar(im1, ax=axs[1])

    im2 = axs[2].imshow(evidencia, cmap='viridis')
    axs[2].set_title(f'Evidencia Visual - Fijación {fixacion}')
    plt.colorbar(im2, ax=axs[2])

    im3 = axs[3].imshow(entropia, cmap='viridis')
    axs[3].set_title(f'Entropía - Fijación {fixacion}')
    plt.colorbar(im3, ax=axs[3])

    for ax in axs:
        ax.axis('off')

    plt.suptitle(f'Mapas para Fijación {fixacion}')
    plt.tight_layout()
    plt.show()

for fix in range(7):
    graficar_mapas_por_fijacion(maps, imagen, fix)
```
